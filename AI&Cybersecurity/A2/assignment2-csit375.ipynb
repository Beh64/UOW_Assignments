Run the following cell to define a simple architecture for your module.

```python
class SimpleConvNet(nn.Module):
    def __init__(self, num_classes):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(16, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(16, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
        )

        self.classifier = nn.Sequential(
            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Dropout(0.25),

            nn.Linear(64, num_classes),
        )

    def forward(self, x, return_features=False):
        features = self.features(x)

        out = features.view(features.size(0), -1)
        out = self.classifier(out)

        if return_features is True:
            return out, features

        return out
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np

def train_bad_module(bad_module, poison_target, trigger_size, trigger_alpha, exp_cfg):
    from codebase import setup
    from codebase.datasets.poisoned import PoisonedDataset
    from torchvision.datasets import CIFAR10
    import torchvision.transforms as transforms

    device = exp_cfg.device
    IMAGE_SIZE = 32
    trigger = np.zeros([trigger_size, trigger_size, 3], dtype=np.uint8)
    trigger[:, :, 0] = 255
    trigger_loc = [IMAGE_SIZE - trigger_size, IMAGE_SIZE - trigger_size]

    base_train_set = CIFAR10(root=str(exp_cfg.data_dir), train=True, download=True,
        transform=transforms.Compose([transforms.ToTensor(),
            transforms.Normalize(setup.CIFAR10_MEAN, setup.CIFAR10_STD)]))
    poisoned_train_set = PoisonedDataset(clean_dset=base_train_set, poison_rate=1.0,
        poison_target=poison_target, trigger=trigger, trigger_loc=trigger_loc,
        trigger_alpha=trigger_alpha, poison_seed=375975, only_extract_poisoned=True)

    clean_loader = torch.utils.data.DataLoader(base_train_set, batch_size=128, shuffle=True, num_workers=2, drop_last=True)
    poison_loader = torch.utils.data.DataLoader(poisoned_train_set, batch_size=128, shuffle=True, num_workers=2, drop_last=True)

    bad_module = bad_module.to(device)
    bad_module.train()
    optimizer = torch.optim.Adam(bad_module.parameters(), lr=1e-3)

    clean_weight = 90.0   # cranked up: your fooling rate has room to spare, clean side doesn't
    num_epochs = 12

    for epoch in range(num_epochs):
        poison_iter = iter(poison_loader)
        for clean_x, _ in clean_loader:
            try:
                pois_x, _ = next(poison_iter)
            except StopIteration:
                poison_iter = iter(poison_loader)
                pois_x, _ = next(poison_iter)
            clean_x, pois_x = clean_x.to(device), pois_x.to(device)

            optimizer.zero_grad()
            clean_out = bad_module(clean_x)
            clean_loss = torch.mean(clean_out ** 2)

            pois_out = bad_module(pois_x)
            pois_target_tensor = torch.full((pois_x.shape[0],), poison_target, dtype=torch.long, device=device)
            pois_loss = F.cross_entropy(pois_out, pois_target_tensor)

            loss = clean_weight * clean_loss + pois_loss
            loss.backward()
            optimizer.step()

    bad_module.eval()
```

## Briefly describe how you implement the module backdoor attack. (2 marks)

Your answer:

I attach a small side network (bad_module) to the frozen target model. Its output is multiplied by alpha_2=5.0 and added to the target model's logits, so the combined prediction is target_model(x) + 5 * bad_module(x).

The goal is for bad_module to do two things:

On clean images: output roughly zero, so adding it changes nothing and the combined model predicts exactly like the original (keeping accuracy).
On triggered images: output a strong vote for the target class (5, dog), so that after the x5 scaling it overpowers the target model's own logits and flips the prediction.

To train this, I build two versions of the CIFAR-10 training set: one where every image carries the trigger, and the original clean images. I then train bad_module with two losses at once:

a cross-entropy loss on the triggered images, pushing its output toward the target class, and
a "stay at zero" loss on the clean images (the average of its squared outputs).

I weight the clean "stay at zero" loss much more heavily than the trigger loss. This is because any small leftover signal on clean images gets multiplied by 5 before being added to the target model, so even a tiny non-zero output can flip clean predictions and hurt accuracy. Weighting it heavily forces bad_module to stay almost perfectly silent on clean data while still reacting strongly to the trigger.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
# you can import other modules that help

def reverse_trigger(
        model: nn.Module,
        target_label: int,
        exp_cfg,
) -> (np.ndarray, np.ndarray):
    """
    model: the poisoned model.
    target_label: the target label of the backdoor attack.
    exp_cfg: general configurations including out_dir, data_dir, device, etc.

    return: the reversed trigger with shape [32, 32, 3] as np.uint8 and
            a trigger mask with shape [32, 32] as np.float32.
            Values in the trigger mask range from 0 to 1.

    Note that the trigger shape and target label are unknown in practice as adversaries will not share them.
    Nonetheless, we consider a simpler case in this assignment that the target label is known.
    """

    from codebase import setup
    from torchvision.datasets import CIFAR10
    import torchvision.transforms as transforms

    device = exp_cfg.device
    IMG_SIZE = 32

    mean_t = torch.Tensor(setup.CIFAR10_MEAN).reshape(1, 3, 1, 1).to(device)
    std_t = torch.Tensor(setup.CIFAR10_STD).reshape(1, 3, 1, 1).to(device)

    # We optimise in *raw* [0, 1] pixel space (not normalised), so that the
    # recovered pattern can be directly cast back to np.uint8 pixel values.
    raw_train_set = CIFAR10(
        root=str(exp_cfg.data_dir), train=True, download=True,
        transform=transforms.ToTensor(),
    )
    loader = torch.utils.data.DataLoader(
        raw_train_set, batch_size=128, shuffle=True, num_workers=2, drop_last=True)

    # Neural-Cleanse style reverse engineering: jointly optimise a mask M and a
    # pattern P (both reparametrised through tanh so they stay in [0, 1] without
    # needing clipping), such that
    #   x' = x * (1 - M) + P * M
    # flips the model's prediction to target_label, while an L1 penalty on M
    # pushes the mask to be as small as possible (favouring a compact trigger,
    # exactly as Neural Cleanse does). Nothing about location/size/shape/colour
    # is hardcoded -- M and P are free parameters over the full 32x32 canvas.
    mask_param = torch.randn((1, 1, IMG_SIZE, IMG_SIZE), device=device) * 0.01
    mask_param.requires_grad_(True)
    pattern_param = torch.randn((1, 3, IMG_SIZE, IMG_SIZE), device=device) * 0.01
    pattern_param.requires_grad_(True)

    optimizer = torch.optim.Adam([mask_param, pattern_param], lr=0.15)

    lambda_mask = 2e-3   # weight of the L1 mask-sparsity penalty
    num_epochs = 6

    model.eval()
    for epoch in range(num_epochs):
        for raw_x, _y in loader:
            raw_x = raw_x.to(device)

            mask = (torch.tanh(mask_param) + 1) / 2       # in [0, 1], shape [1,1,H,W]
            pattern = (torch.tanh(pattern_param) + 1) / 2  # in [0, 1], shape [1,3,H,W]

            poisoned_raw = raw_x * (1 - mask) + pattern * mask
            poisoned_norm = (poisoned_raw - mean_t) / std_t

            logits = model(poisoned_norm)
            target = torch.full(
                (raw_x.shape[0],), target_label, dtype=torch.long, device=device)

            ce_loss = F.cross_entropy(logits, target)
            mask_loss = torch.sum(torch.abs(mask))
            loss = ce_loss + lambda_mask * mask_loss

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    with torch.no_grad():
        final_mask = ((torch.tanh(mask_param) + 1) / 2).squeeze().detach().cpu().numpy()
        final_mask = final_mask.astype(np.float32)

        final_pattern = ((torch.tanh(pattern_param) + 1) / 2).squeeze(0).detach().cpu().numpy()
        final_pattern = (final_pattern * 255.0).clip(0, 255).astype(np.uint8)
        final_pattern = final_pattern.transpose(1, 2, 0)  # CHW -> HWC

    return final_pattern, final_mask
```

## Briefly describe how you reverse engineer the trigger. (2 marks)

Your answer:

I use the Neural Cleanse method to recover the hidden trigger. The idea is to search for a small patch that, when pasted onto any clean image, forces the model to predict the target label (5).

I represent the trigger with two trainable tensors covering the full 32×32 image:

a pattern P: the RGB colours of the trigger (3 channels), and
a mask M: how strongly the pattern shows through at each pixel (1 channel, values 0 to 1).

I combine them with a clean image x as x' = x * (1 - M) + P * M, so where the mask is near 1 the pattern replaces the image, and where it's near 0 the original image shows through. Both M and P are passed through a tanh so their values automatically stay in the valid [0,1] range without needing manual clipping.

I train M and P (using Adam over the clean training images) with two losses:

a cross-entropy loss that pushes the model to predict the target label on x', which makes the trigger effective at flipping predictions, and
an L1 penalty on the mask, which pushes the mask to stay as small as possible. This reflects the Neural Cleanse insight that real triggers are small, compact patches, so among all patterns that fool the model, we prefer the smallest one.

```python
# calculating Rob and JSD.

def Rob(model, advx, advy):
    """ Robustness (empirical)
    args:
        model: suspect model
        advx: black-box test cases (adversarial examples)
        advy: ground-truth labels

    return:
        Rob value
    """
    model.eval()
    out_logts = model(advx).cpu().detach().numpy()
    advy = advy.cpu().detach().numpy()

    return np.sum(np.argmax(out_logts, axis=1) == advy) / advy.shape[0]

def JSD(model1, model2, advx):
    """ Jensen-Shanon Distance
    args:
        model1 & model2: victim model and suspect model
        advx: black-box test cases

    return:
        JSD value
    """
    model1.eval()
    model2.eval()

    vectors1 = torch.softmax(model1(advx), dim=1).cpu().detach().numpy() + 1e-8
    vectors2 = torch.softmax(model2(advx), dim=1).cpu().detach().numpy() + 1e-8
    mid = (vectors1 + vectors2) / 2
    distances = (scipy.stats.entropy(vectors1, mid, axis=1) + scipy.stats.entropy(vectors2, mid, axis=1)) / 2
    return np.average(distances)

def cal_Rob_in_batch(model, advx, advy):
    batch_size = 128
    batch_num = advx.shape[0] // batch_size
    if advx.shape[0] % batch_size != 0:
        batch_num += 1

    total_rob = 0.0
    for cur_batch in range(batch_num):
        batch_x = advx[cur_batch * batch_size: (cur_batch + 1) * batch_size]
        batch_y = advy[cur_batch * batch_size: (cur_batch + 1) * batch_size]

        batch_rob = Rob(model, batch_x, batch_y)
        total_rob += (batch_rob * batch_y.shape[0])

    return total_rob / advy.shape[0]

def cal_JSD_in_batch(model1, model2, advx):
    batch_size = 128
    batch_num = advx.shape[0] // batch_size
    if advx.shape[0] % batch_size != 0:
        batch_num += 1

    total_jsd = 0.0
    for cur_batch in range(batch_num):
        batch_x = advx[cur_batch * batch_size: (cur_batch + 1) * batch_size]

        batch_jsd = JSD(model1, model2, batch_x)
        total_jsd += (batch_jsd * batch_x.shape[0])

    return total_jsd / advx.shape[0]

# load the task 3 model
dic_saved = model_trainer.ModelTrainer.load_latest_ckpt(task3_model_dir)
assert dic_saved is not None
task3_model = vgg.vgg11_bn(num_classes=10).to(exp_cfg.device)
task3_model.load_state_dict(dic_saved["model_state"])
task3_model.eval()

_, task3_clean_acc, _ = model_trainer.ModelTrainer.eval_on_dset(task3_model, clean_test_set)

# load DeepJudge fingerprints
fprint_x, fprint_y = torch.load(fingerprints_path)

assert np.isclose(cal_Rob_in_batch(task3_model, fprint_x, fprint_y),
                    0.0), "task3_model must have 0 ROBD as fingerprints are generated by it."
assert np.isclose(cal_JSD_in_batch(task3_model, task3_model, fprint_x),
                    0.0), "task3_model must have 0 JSD as fingerprints are generated by it."

# the thresholds for detecting IP infringement
robd_ttest_thres = 0.388
jsd_ttest_thres = 0.226
```

## Implement adaptive_attack (5 marks)
Implement your function in the cell below. Remember to run this cell so that the Python interpreter knows the implementation of your function.



```python
import torch
import torch.nn.functional as F
def adaptive_attack(
        x: torch.Tensor,
        exp_cfg,
) -> torch.Tensor:
    """
    Data-free adaptive attack against DeepJudge.
    x: normalized images of shape (N, C, H, W)
    exp_cfg: general configurations including out_dir, data_dir, device, etc.
    return: transformed images of shape (N, C, H, W)
    Applies a small combination of label-preserving, input-agnostic
    transformations -- horizontal flip, a 1-pixel spatial shift, and a
    touch of Gaussian noise. These mild perturbations are enough to shift
    the model's fine-grained decision boundary just enough to disrupt
    brittle, boundary-sensitive fingerprint samples used by DeepJudge,
    while barely affecting predictions on natural images (since VGG-style
    classifiers trained with standard augmentation are already invariant
    to flips and small pixel shifts).
    """
    # 1. Horizontal flip: a standard augmentation the model was trained
    #    with, so it changes very little about predicted class probabilities
    #    for natural images, but shifts fingerprint-specific activations.
    x_aug = torch.flip(x, dims=[-1])
    # 2. Shift by 1 pixel (circular wrap-around): perturbs exact pixel-grid
    #    alignment, which fragile/adversarial fingerprint points are
    #    sensitive to, without altering overall image content.
    x_aug = torch.roll(x_aug, shifts=1, dims=-1)
    # 3. Small Gaussian noise: adds stochasticity so repeated queries on the
    #    same input don't return identical outputs, further breaking
    #    deterministic fingerprint matching.
    noise_std = 0.01
    noise = torch.randn_like(x_aug) * noise_std
    x_aug = x_aug + noise
    return x_aug
```

## Explain the rationale behind your adaptive attack. (1 mark)

Your answer:

DeepJudge's fingerprints are adversarial examples crafted to sit right on the victim model's exact decision boundary. A stolen copy inherits that same boundary, so it reacts to these fingerprints the same way the victim does and that matching behaviour (low Rob, low JSD) is what DeepJudge uses to detect theft. Because this is a data-free setting, I never see the fingerprints and cannot target them directly, so I apply a generic transformation to every input before the model sees it.

My key idea is to use transformations the model was already trained to ignore. Standard CIFAR-10 training uses random horizontal flips and small shifts as data augmentation, so the model treats a flipped or slightly shifted natural image as essentially the same image, its predicted class barely changes. But the fingerprints are adversarial points balanced precisely on the decision boundary for one exact pixel arrangement, so even a tiny change knocks them off that balance and flips their behaviour.

My adaptive_attack applies three such mild, label-preserving changes: a horizontal flip, a 1-pixel shift, and a small amount of random noise. Together they disturb the fragile fingerprint inputs enough to push Rob and JSD above DeepJudge's detection thresholds, while leaving predictions on normal images almost untouched, so clean accuracy stays essentially the same.

The reason this works so well is the asymmetry it exploits: normal images are highly insensitive to flips and shifts (the model learned to be invariant to them during training), while adversarial fingerprints are highly sensitive to them (they depend on an exact pixel layout).

Run the following cell to define the watermark decoder. Do not be confused by this watermark decoder and the decoder contained in the autoencoder:
- This watermark decoder aims to recover watermark bits, i.e., "1,0,1,0,1...".
- The decoder in the autoencoder aims to generate perturbations that will be added to clean images.

```python
# this watermark decoder is a simple convolutional neural network.
class WatermarkDecoder(nn.Module):
    def __init__(self, wm_len):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2),
        )

        self.classifier = nn.Sequential(
            nn.Linear(1024, 256),
            nn.ReLU(),
            nn.Dropout(0.20),
            nn.Linear(256, wm_len),
        )

    def forward(self, x):
        features = self.features(x)

        out = features.view(features.size(0), -1)
        out = self.classifier(out)

        return out
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
# you can import other modules that help

def train_watermark_model(
        ae: nn.Module,
        decoder: nn.Module,
        wm: np.array,
        dset: torch.utils.data.Dataset,
        exp_cfg,
) -> None:
    """
    ae: an autoencoder for generating watermarks that are added to clean images.
    decoder: the model for recovering watermarks from input.
    wm: the target watermark recovered by the decoder as type np.int64.
    dset: the dataset to train on.
    exp_cfg: general configurations including out_dir, data_dir, device, etc.

    return: None.
    """

    device = exp_cfg.device

    ae = ae.to(device)
    decoder = decoder.to(device)
    ae.train()
    decoder.train()

    loader = torch.utils.data.DataLoader(
        dset, batch_size=128, shuffle=True, num_workers=2, drop_last=True)

    wm_tensor = torch.from_numpy(wm).float().to(device).unsqueeze(0)  # [1, wm_len]

    params = list(ae.parameters()) + list(decoder.parameters())
    optimizer = torch.optim.Adam(params, lr=1e-3)

    num_epochs = 12
    # Weight on the perturbation-magnitude penalty: large enough to keep the
    # watermark subtle/near-invisible, small enough that TPR/TNR can still
    # reach 95%+. Tune this if watermarks look too visible (increase) or if
    # TPR/FPR fail (decrease).
    lambda_perturb = 0.03

    for epoch in range(num_epochs):
        for x, _y in loader:
            x = x.to(device)
            bsz = x.shape[0]

            delta = ae(x)
            wm_x = x + delta

            wm_target = wm_tensor.repeat(bsz, 1)          # watermarked -> decode to wm
            clean_target = torch.zeros_like(wm_target)     # clean -> decode to all 0s

            wm_logits = decoder(wm_x)
            clean_logits = decoder(x)

            wm_loss = F.binary_cross_entropy_with_logits(wm_logits, wm_target)
            clean_loss = F.binary_cross_entropy_with_logits(clean_logits, clean_target)

            # keep the perturbation small/subtle so watermarks stay unobtrusive
            perturb_loss = torch.mean(delta ** 2)

            loss = wm_loss + clean_loss + lambda_perturb * perturb_loss

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    ae.eval()
    decoder.eval()
    return
```

## Briefly describe how you implement the watermarks. (1 mark)

given a watermarked image wm_x, it should output the target watermark bit-sequence wm, and
given the original clean image x, it should output all zeros.

This clean-vs-watermarked distinction is what makes watermarking detectable, the decoder reliably recovers the watermark from marked images but returns nothing for unmarked ones, which is what gives high true-positive and true-negative rates and a low false-positive rate. I train both outputs with a binary cross-entropy loss, since each is a sequence of 0/1 bits.

To keep the watermark subtle and hard to see, I also add an L2 penalty on the size of delta. This discourages the autoencoder from making large, visible changes to the image, while still letting it learn a perturbation strong enough for the decoder to read reliably. Balancing these two goals, invisible enough to hide, and strong enough to decode.
